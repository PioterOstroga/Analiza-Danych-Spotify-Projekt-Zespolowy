```python
!pip install -q missingno
import pandas as pd
import missingno as msno
import matplotlib.pyplot as plt
from google.colab import files

wgrane = files.upload()
nazwa = list(wgrane.keys())[0]
print("Wczytuję:", nazwa)

# Dynamiczne dopasowanie metody wczytywania do typu pliku
if nazwa.endswith('.xlsx') or nazwa.endswith('.xls'):
    df = pd.read_excel(nazwa)
else:
    df = pd.read_csv(nazwa)

print(df.shape)
```

```python
# braki w liczbach
braki = df.isna().sum()
tabela = pd.DataFrame({
    "liczba_brakow": braki,
    "procent": (braki / len(df) * 100).round(2)
})
tabela = tabela[tabela["liczba_brakow"] > 0].sort_values("liczba_brakow", ascending=False)

print(f"Liczba wierszy: {len(df)}")
print(f"Łącznie braków: {braki.sum()}")
display(tabela)

# wykres braków
msno.matrix(df, figsize=(14, 6), fontsize=10, sparkline=False)
plt.title("Shadowmapa braków danych", fontsize=14)
plt.show()
```

```python
def znajdz_dziwne_dane(df):
    # granice dla ceny odstającej (metoda IQR)
    q1, q3 = df["Price"].quantile([0.25, 0.75])
    iqr = q3 - q1

    reguly = pd.DataFrame({
        "rok_budowy_niemozliwy":       (df["YearBuilt"] < 1840) | (df["YearBuilt"] > 2018),
        "pow_budynku_0":               df["BuildingArea"] == 0,
        "pow_budynku_ponizej_20m2":    (df["BuildingArea"] > 0) & (df["BuildingArea"] < 20),
        "dom_bez_dzialki":             (df["Type"] == "h") & (df["Landsize"] == 0),
        "brak_lazienki":               df["Bathroom"] == 0,
        "brak_sypialni":               df["Bedroom2"] == 0,
        "sypialni_wiecej_niz_pokoi":   df["Bedroom2"] > df["Rooms"] + 2,
        "budynek_wiekszy_niz_dzialka": (df["Type"] == "h") & (df["Landsize"] > 0)
                                       & (df["BuildingArea"] > df["Landsize"]),
        "cena_odstajaca_IQR":          (df["Price"] < q1 - 1.5 * iqr) | (df["Price"] > q3 + 1.5 * iqr),
        "duplikat":                    df.duplicated(),
    })
    return reguly

dziwne = znajdz_dziwne_dane(df)

# True/False dla pierwszych wierszy
display(dziwne.head(10))

# ile True w każdej regule
podsumowanie = dziwne.sum().sort_values(ascending=False).rename("liczba_True")
display(podsumowanie)

# wiersze, które łamią choć jedną regułę
df["czy_podejrzany"] = dziwne.any(axis=1)
print("Podejrzanych wierszy:", df["czy_podejrzany"].sum())

# podgląd przykładowej reguły
display(df.loc[dziwne["rok_budowy_niemozliwy"], ["Suburb", "Type", "YearBuilt", "Price"]])
```
